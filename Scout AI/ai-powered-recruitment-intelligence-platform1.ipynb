```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
# Generate embeddings for the first 5 resumes
resume_embeddings = model.encode(
    resume_df["resume_text"].head(5).tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

# Generate embeddings for the first 5 jobs
job_embeddings = model.encode(
    skills_df["job_text"].head(5).tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

print("Resume embeddings shape:", resume_embeddings.shape)
print("Job embeddings shape:", job_embeddings.shape)
```

```python
similarity_matrix = cosine_similarity(
    resume_embeddings,
    job_embeddings
)

print(similarity_matrix)
```

```python
# Generate embeddings for all jobs
job_embeddings = model.encode(
    skills_df["job_text"].tolist(),
    convert_to_numpy=True,
    show_progress_bar=True
)

print(job_embeddings.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_jobs(resume_text, top_k=5):
    # Convert resume into an embedding
    resume_embedding = model.encode([resume_text], convert_to_numpy=True)

    # Compare with all job embeddings
    similarities = cosine_similarity(resume_embedding, job_embeddings)[0]

    # Get indices of the top matching jobs
    top_indices = similarities.argsort()[-top_k:][::-1]

    recommendations = skills_df.iloc[top_indices][
        ["job_title", "category", "job_skill_set"]
    ].copy()

    recommendations["match_score"] = similarities[top_indices]

    return recommendations
```

```python
import ast

def extract_resume_skills(skill_text):
    if pd.isna(skill_text) or skill_text == "":
        return []

    try:
        return [skill.lower().strip() for skill in ast.literal_eval(skill_text)]
    except:
        return []

# Test on the first resume
resume_skills = extract_resume_skills(resume_df.iloc[0]["skills"])

print(resume_skills)
```

```python
# Resume skills
resume_skills = set(extract_resume_skills(resume_df.iloc[0]["skills"]))

# Recommended job skills
job_skills = {
    skill.lower().strip()
    for skill in skills_df.iloc[897]["job_skill_set"]
}

print("Resume Skills:", len(resume_skills))
print("Job Skills:", len(job_skills))
```

```python
matched_skills = resume_skills.intersection(job_skills)
missing_skills = job_skills.difference(resume_skills)

print("=" * 60)
print("Matched Skills")
print("=" * 60)
print(sorted(matched_skills))

print("\n")

print("=" * 60)
print("Missing Skills")
print("=" * 60)
print(sorted(missing_skills))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def recommend_jobs_with_skill_gap(resume_index, top_k=5):
    # Get resume text
    resume_text = resume_df.iloc[resume_index]["resume_text"]

    # Generate embedding
    resume_embedding = model.encode([resume_text], convert_to_numpy=True)

    # Compare with all jobs
    similarities = cosine_similarity(resume_embedding, job_embeddings)[0]

    # Get Top-K jobs
    top_indices = similarities.argsort()[-top_k:][::-1]

    # Resume skills
    resume_skills = set(
        extract_resume_skills(
            resume_df.iloc[resume_index]["skills"]
        )
    )

    recommendations = []

    for idx in top_indices:

        job_skills = {
            skill.lower().strip()
            for skill in skills_df.iloc[idx]["job_skill_set"]
        }

        matched = sorted(resume_skills & job_skills)
        missing = sorted(job_skills - resume_skills)

        recommendations.append({
            "job_title": skills_df.iloc[idx]["job_title"],
            "category": skills_df.iloc[idx]["category"],
            "match_score": round(float(similarities[idx]), 4),
            "matched_skills": matched,
            "missing_skills": missing
        })

    return recommendations
```

```python
results = recommend_jobs_with_skill_gap(0)

for job in results:
    print("=" * 80)
    print("Job:", job["job_title"])
    print("Category:", job["category"])
    print("Match Score:", job["match_score"])

    print("\nMatched Skills:")
    print(job["matched_skills"])

    print("\nMissing Skills:")
    print(job["missing_skills"])
```

```python
# Keep only useful columns
interview_df = interview_df[[
    "Question",
    "Answer",
    "Category",
    "Difficulty"
]].copy()

print(interview_df.head())
```

```python
def recommend_interview_questions(category=None, difficulty=None, top_k=5):

    df = interview_df.copy()

    if category:
        df = df[df["Category"].str.lower() == category.lower()]

    if difficulty:
        df = df[df["Difficulty"].str.lower() == difficulty.lower()]

    return df.head(top_k)
```

```python
recommend_interview_questions(
    category="General Programming",
    top_k=5
)
```

```python
def scout_ai(resume_index, top_k=5):
    # Get recommendations
    job_results = recommend_jobs_with_skill_gap(resume_index, top_k)

    print("=" * 100)
    print("SCOUT AI CAREER REPORT")
    print("=" * 100)

    for i, job in enumerate(job_results, start=1):

        print(f"\n🏆 Recommendation #{i}")
        print("-" * 100)

        print("Job Title :", job["job_title"])
        print("Category  :", job["category"])
        print("Match     :", f"{job['match_score']:.2%}")

        print("\n✅ Matching Skills")
        if job["matched_skills"]:
            print(", ".join(job["matched_skills"]))
        else:
            print("None")

        print("\n❌ Missing Skills")
        if job["missing_skills"]:
            print(", ".join(job["missing_skills"]))
        else:
            print("None")

    print("\n" + "=" * 100)
    print("Recommended Interview Questions")
    print("=" * 100)

    questions = recommend_interview_questions(top_k=5)

    for i, row in questions.iterrows():
        print(f"\nQ{i+1}: {row['Question']}")
        print(f"Difficulty: {row['Difficulty']}")
        print(f"Category: {row['Category']}")
```
